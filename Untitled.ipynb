```python
import pandas as pd
from pycaret.classification import *
```

```python
# 1. Load the dataset
data = pd.read_csv('Data/GlaucomaM.csv')

# Preview the data
print(data.sample(5))
```

```python
# 2. Initialize the PyCaret Environment
# 'Class' is the target column based on the GlaucomaM dataset structure
s = setup(
    data=data, 
    target='Class', 
    session_id=123, 
    normalize=True,           # Normalize features
    transformation=True,      # Apply power transformations for skewness
    feature_selection=True,   # Automatically select important features
    fold=5                    # 5-fold cross-validation
)
```

```python
# 3. Compare all available models to find the best performer
best_model = compare_models()
```

```python
# 4. Analyze the best model (e.g., Confusion Matrix, AUC, Feature Importance)
evaluate_model(best_model)
```

```python
# 5. Tune the best model hyperparameters
tuned_best_model = tune_model(best_model)
```

```python
# 6. Finalize the model for deployment
final_model = finalize_model(tuned_best_model)
```

```python
# 7. Save the model pipeline for future predictions
save_model(final_model, 'glaucoma_classification_model')
print("Model built, optimized, and saved successfully!")
```

```python
# ==========================================
# 4. Generate SHAP Values & Plots using PyCaret
# ==========================================

# A. Summary Plot (Beeswarm Plot)
# Shows the top features, their impact magnitude, and the direction of the relationship (high/low feature values vs. predictions)
print("Generating SHAP Summary Plot...")
s.interpret_model(best_model, plot="summary")
```

```python
# B. Correlation / Bar Plot
# Shows the global importance of each feature based on the average magnitude of their SHAP values
print("Generating SHAP Feature Importance Bar Plot...")
s.interpret_model(best_model, plot="correlation") 
# Note: In PyCaret, plot="reason" or plot="summary" handles the main SHAP visualizations. 
# You can also use plot="decorator" or individual feature dependency plots.
```

```python
# C. Individual Explanation (Waterfall / Reason Plot for a single instance)
# Explains a specific prediction (e.g., the very first row in your dataset)
print("Generating Individual Prediction Explanation...")
s.interpret_model(best_model, plot="reason", observation=0)
```
