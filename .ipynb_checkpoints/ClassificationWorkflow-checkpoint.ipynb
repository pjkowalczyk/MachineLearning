## Data  

**Connectionist Bench** (Sonar, Mines vs. Rocks)
- Binary classification (distinguishing between sonar signals bouncing off a metal cylinder vs. a rough cylindrical rock).
- 60 purely numeric continuous features representing the energy within a particular frequency band. ~208 instances.

```python
import sklearn
# X = sklearn.datasets.fetch_openml(name='sonar', version = 1).data
# y = sklearn.datasets.fetch_openml(name='sonar', version = 1).target

# Load the Sonar Dataset (Sonar, Mines vs. Rocks)
# Target classes: 'R' for Rock, 'M' for Mine
sonar = fetch_openml(name='sonar', version=1, as_frame=True)
X = sonar.data
# Convert target labels to binary (1 for Mine 'M', 0 for Rock 'R')
y = (sonar.target == 'Mine').astype(int)
```

### Remove low variance features

```python
variance_threshold = 0.01

selector = VarianceThreshold(threshold = variance_threshold)
selector.fit(X)
    
# Get retained feature names
retained_features = X.columns[selector.get_support()]
removed_low_var = set(X.columns) - set(retained_features)
    
X_low_var_removed = pd.DataFrame(selector.transform(X), columns = retained_features)
    
print(f"\n[Step 1] Low variance features removed ({len(removed_low_var)}): {list(removed_low_var)}")
print(f"\nShape after low-variance filtering: {X_low_var_removed.shape}")
```

### Remove highly correlated features

```python
correlation_threshold = 0.85

# Compute absolute correlation matrix
corr_matrix = X_low_var_removed.corr().abs()
    
# Select upper triangle of correlation matrix
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k = 1).astype(bool))
    
# Find features with correlation greater than the threshold
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > correlation_threshold)]
    
X_final = X_low_var_removed.drop(columns=to_drop)
    
print(f"\n[Step 2] Highly correlated features removed ({len(to_drop)}): {to_drop}")
print(f"\nFinal features dataframe shape: {X_final.shape}")
```

```python
# Split into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size = 0.2, random_state = 42, stratify = y
)
```

## Classification

```python
import warnings
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    cohen_kappa_score
)

# Machine Learning Classifiers
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier
)
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

# Third-party Boosting Libraries
from catboost import CatBoostClassifier

# Suppress non-critical warnings
warnings.filterwarnings('ignore')

# 1. Load the Sonar Dataset (UCI Sonar dataset via OpenML)
print("Loading Sonar dataset...")
sonar = fetch_openml(name='sonar', version=1, as_frame=True)
X = sonar.data
y = sonar.target

# Encode labels ('R' -> 0, 'M' -> 1)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# 2. Split into Train and Test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded
)

# 3. Scale Features (Essential for distance/gradient-based algorithms)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Define all 17 Classifiers
classifiers = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM (Linear)": SVC(kernel='linear', random_state=42),
    "SVM (RBF / Radial)": SVC(kernel='rbf', random_state=42),
    "Gaussian Process": GaussianProcessClassifier(random_state=42),
    "MLP Classifier": MLPClassifier(max_iter=1000, random_state=42),
    "Ridge Classifier": RidgeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Quadratic Discriminant Analysis": QuadraticDiscriminantAnalysis(),
    "AdaBoost Classifier": AdaBoostClassifier(random_state=42),
    "Gradient Boosting Classifier": GradientBoostingClassifier(random_state=42),
    "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
    "Extra Trees Classifier": ExtraTreesClassifier(random_state=42),
    "CatBoost": CatBoostClassifier(random_state=42, verbose=0)
}

# 5. Train, Evaluate, and Collect Metrics
results = []

for name, clf in classifiers.items():
    # Fit model on scaled training data
    clf.fit(X_train_scaled, y_train)
    
    # Predict on test set
    y_pred = clf.predict(X_test_scaled)
    
    # Calculate Metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    kappa = cohen_kappa_score(y_test, y_pred)
    
    # Confusion Matrix Components
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    
    results.append({
        "Classifier": name,
        "Accuracy": round(acc, 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1 Score": round(f1, 4),
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "kappa": round(kappa, 4)
    })

# 6. Store and Display Results
results_df = pd.DataFrame(results)

# Sort DataFrame by Accuracy in descending order
results_df = results_df.sort_values(by="kappa", ascending=False).reset_index(drop=True)

print("\nEvaluation Results:")
print(results_df.to_string(index=False))
```

```python
import warnings
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import VarianceThreshold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    cohen_kappa_score
)

# Machine Learning Classifiers
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier
)
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)
from catboost import CatBoostClassifier

# Suppress non-critical warnings
warnings.filterwarnings('ignore')

# 1. Load the Sonar Dataset (UCI Sonar dataset via OpenML)
print("Loading Sonar dataset...")
sonar = fetch_openml(name='sonar', version=1, as_frame=True)
X = sonar.data
y = sonar.target

###
# remove low variance features
###
variance_threshold = 0.01

selector = VarianceThreshold(threshold = variance_threshold)
selector.fit(X)
    
# Get retained feature names
retained_features = X.columns[selector.get_support()]
removed_low_var = set(X.columns) - set(retained_features)
    
X_low_var_removed = pd.DataFrame(selector.transform(X), columns = retained_features)

###
# remove highly correlated features
###

correlation_threshold = 0.85

# Compute absolute correlation matrix
corr_matrix = X_low_var_removed.corr().abs()
    
# Select upper triangle of correlation matrix
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k = 1).astype(bool))
    
# Find features with correlation greater than the threshold
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > correlation_threshold)]
    
X = X_low_var_removed.drop(columns=to_drop)

# Encode labels ('R' -> 0, 'M' -> 1)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# 2. Train-Test Split (70/30 stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded
)

# 3. Standardize Features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Define Models and Hyperparameter Grids for GridSearchCV
models_and_params = {
    "Logistic Regression": (
        LogisticRegression(random_state=42),
        {'C': [0.01, 0.1, 1.0, 10.0], 'solver': ['lbfgs', 'liblinear']}
    ),
    "K-Nearest Neighbors": (
        KNeighborsClassifier(),
        {'n_neighbors': [3, 5, 7, 9], 'weights': ['uniform', 'distance']}
    ),
    "Naive Bayes": (
        GaussianNB(),
        {'var_smoothing': np.logspace(0, -9, num=5)}
    ),
    "Decision Tree": (
        DecisionTreeClassifier(random_state=42),
        {'max_depth': [None, 3, 5, 10], 'criterion': ['gini', 'entropy']}
    ),
    "SVM (Linear)": (
        SVC(kernel='linear', random_state=42),
        {'C': [0.1, 1.0, 10.0]}
    ),
    "SVM (RBF / Radial)": (
        SVC(kernel='rbf', random_state=42),
        {'C': [0.1, 1.0, 10.0], 'gamma': ['scale', 'auto', 0.1]}
    ),
    "Gaussian Process": (
        GaussianProcessClassifier(random_state=42),
        {'max_iter_predict': [100, 200]}
    ),
    "MLP Classifier": (
        MLPClassifier(max_iter=1000, random_state=42),
        {'hidden_layer_sizes': [(50,), (100,)], 'alpha': [0.0001, 0.01]}
    ),
    "Ridge Classifier": (
        RidgeClassifier(random_state=42),
        {'alpha': [0.1, 1.0, 10.0]}
    ),
    "Random Forest": (
        RandomForestClassifier(random_state=42),
        {'n_estimators': [50, 100], 'max_depth': [None, 5, 10]}
    ),
    "Quadratic Discriminant Analysis": (
        QuadraticDiscriminantAnalysis(),
        {'reg_param': [0.0, 0.1, 0.5]}
    ),
    "AdaBoost Classifier": (
        AdaBoostClassifier(random_state=42),
        {'n_estimators': [50, 100], 'learning_rate': [0.1, 1.0]}
    ),
    "Gradient Boosting Classifier": (
        GradientBoostingClassifier(random_state=42),
        {'n_estimators': [50, 100], 'learning_rate': [0.05, 0.1], 'max_depth': [3, 5]}
    ),
    "Linear Discriminant Analysis": (
        LinearDiscriminantAnalysis(),
        {'solver': ['svd', 'lsqr']}
    ),
    "Extra Trees Classifier": (
        ExtraTreesClassifier(random_state=42),
        {'n_estimators': [50, 100], 'max_depth': [None, 5, 10]}
    ),
    "CatBoost": (
        CatBoostClassifier(random_state=42, verbose=0),
        {'iterations': [50, 100], 'depth': [4, 6], 'learning_rate': [0.05, 0.1]}
    )
}

# 5. Run GridSearchCV, Evaluate, and Collect Metrics
results = []

print("Running GridSearchCV for all classifiers...\n")
for name, (model, param_grid) in models_and_params.items():
    print(f"Tuning {name}...")
    
    # Setup 5-fold cross-validation GridSearch based on F1 Score
    grid_search = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        cv=5,
        scoring='f1',
        n_jobs=-1
    )
    
    # Fit GridSearch on scaled training data
    grid_search.fit(X_train_scaled, y_train)
    best_model = grid_search.best_estimator_
    
    # Predict on unseen test set
    y_pred = best_model.predict(X_test_scaled)
    
    # Calculate performance metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    kappa = cohen_kappa_score(y_test, y_pred)
    
    # Extract confusion matrix values
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    
    results.append({
        "Classifier": name,
        "Best Parameters": str(grid_search.best_params_),
        "Acc": round(acc, 4),
        "Prec": round(prec, 4),
        "Rec": round(rec, 4),
        "F1": round(f1, 4),
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "kappa": round(kappa, 4)
    })

# 6. Format and Display Final DataFrame
results_df = pd.DataFrame(results)

# Sort by kappa in descending order
results_df = results_df.sort_values(by="kappa", ascending=False).reset_index(drop=True)

print("\n" + "="*80)
print("FINAL EVALUATION RESULTS (GRIDSEARCHCV TUNED)")
print("="*80)
print(results_df.to_string(index=False))
```
